```python
import matplotlib.pyplot as plt

df =pd.read_csv("cybersecurity_threat_detection_logs.csv", nrows=10000)


blocked = df[df["action"] == "blocked"]

counts = blocked["protocol"].value_counts()
plt.figure()
plt.bar(counts.index, counts.values)
plt.title("Protocols and threats detected")
plt.xlabel("Protocols")
plt.ylabel("threat_label")
plt.show()

print(counts)
print("  ")
print("Insights:")
print("TCP is the most frequently observed protocol , suggesting that most network communications rely on connection-oriented services.")
print("The high volume of TCP traffic makes TCP-based services a primary area for security monitoring.")
print("Significant HTTP and HTTPS activity may require closer inspection for web-based threats and suspicious application traffic.")
print("The presence of substantial UDP traffic warrants investigation to determine whether it represents legitimate services or potentially suspicious network behavior.")
```

```python
import matplotlib.pyplot as plt

df =pd.read_csv("cybersecurity_threat_detection_logs.csv", nrows=100)


blocked = df[df["action"] == "blocked"]

counts = blocked["protocol"].value_counts()
plt.figure()
plt.bar(counts.index, counts.values)
plt.title("Blocked Protocols")
plt.xlabel("Protocols")
plt.ylabel("Number Blocked")
plt.show()

#print(counts)
print("Insights:")
print("UDP generated a high number of blocked connections in the network.")
print("The high block rate may indicate attempts to exploit UDP-based services.")
print("Excessive UDP traffic can be associated with network scanning or denial-of-service (DoS) attacks.")
print("Blocking UDP traffic helps reduce the risk of amplification attacks.")
```
